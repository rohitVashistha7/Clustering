#1.	What is unsupervised learning in the context of machine learning?

-> Unsupervised learning in machine learning refers to algorithms that learn from unlabeled data without human supervision. Unlike supervised learning, where models are trained on labeled datasets to predict outcomes, unsupervised learning models identify patterns, structures, or groupings in raw data on their own.


# 2.	How does K—Means clustering algorithm work*
-> K-means clustering is an unsupervised learning algorithm that groups data points into a predefined number of clusters (k) by iteratively minimizing the distance within clusters and maximizing the distance between clusters.

How K-means works:

1.	Choose k centroids: Randomly select k points as initial cluster centers (centroids).
2.	Assign points to clusters: Calculate the distance (usually Euclidean) from each data point to each centroid and assign the point to the nearest centroid.
3.	Update centroids: Compute the new centroid of each cluster by taking the mean of all points assigned to that cluster.
4.	Repeat: Reassign points based on updated centroids and recalculate centroids until the centroids no longer move significantly or a maximum number of iterations is reached.

The goal is to minimize the sum of squared distances between points and their respective cluster centroids, achieving compact and well-separated clusters


#3.	Explain the concept of a dendrogram in hierarchical clustering.
-> A dendrogram is a tree-like diagram that visually represents the hierarchical relationships between objects or data points, commonly used as the output of hierarchical clustering algorithms

Key aspects of a dendrogram:

•	Structure: It shows how individual data points or clusters are merged step-by-step into larger clusters, forming a nested hierarchy.

•	Axes: The x-axis typically represents the individual data points, while the y-axis (height) indicates the distance or dissimilarity at which clusters are joined.

•	Interpretation: The height at which two clusters merge reflects their similarity-the lower the height, the more similar the clusters are. For example, clusters joined at a small height are more similar than those joined higher up.

•	Usage: By cutting the dendrogram at a chosen height (threshold), you can decide the number of clusters in the data. This is done by drawing a horizontal line and counting how many vertical lines it crosses.

•	Information: The dendrogram summarizes the distance matrix used in clustering, but some information loss can occur, especially for complex data not satisfying certain mathematical properties.


# 4.	What is the main difference between K-Means and Hierarchical Clustering*
-> The main difference between K-Means and Hierarchical Clustering lies in how they form clusters and handle the number of clusters:

K-Means Clustering partitions data into a fixed number of clusters (k) specified beforehand. It iteratively assigns points to clusters based on proximity to centroids and updates these centroids until convergence. It assumes spherical, similarly sized clusters and is computationally efficient, making it suitable for large datasets.

Hierarchical Clustering builds a hierarchy of clusters either by a bottom-up (agglomerative) or top-down (divisive) approach, creating a dendrogram that shows cluster relationships at multiple levels. It does not require specifying the number of clusters in advance; instead, you can choose clusters by cutting the dendrogram at a desired level. It can handle clusters of varying shapes and sizes but is computationally expensive and better suited for smaller datasets.

# 5.	What are the advantages of DBSCAN over K-Means?
DBSCAN has several advantages over K-Means clustering:


•	Ability to find clusters of arbitrary shapes: DBSCAN can discover clusters with complex, irregular shapes, while K-Means assumes clusters are spherical and tends to perform poorly on non-globular clusters.

•	No need to specify the number of clusters beforehand: DBSCAN automatically determines the number of clusters based on data density, whereas K-Means requires the number of clusters (k) to be set in advance.

•	Robustness to noise and outliers: DBSCAN identifies noise points and excludes them from clusters, making it more robust to outliers. K-Means assigns every point to a cluster, which can distort cluster centers if outliers are present.

•	Handles clusters of varying sizes and densities: DBSCAN can detect clusters with different densities and sizes, while K-Means struggles when cluster sizes vary significantly.

However, DBSCAN may perform poorly on high-dimensional sparse data or when densities vary dramatically within the dataset


# 6.	When would you use Silhouette Score in clustering
-> The Silhouette Score is used in clustering to evaluate how well data points have been grouped into clusters by measuring both cohesion (how close points are within the same cluster) and separation (how distinct clusters are from each other).

When to use Silhouette Score:

•	Choosing the optimal number of clusters: By calculating the average silhouette score for different numbers of clusters, you can identify the cluster count that yields the best-defined and most separated clusters, often used with algorithms like K-Means.

•	Validating cluster quality: It provides a quantitative measure of how appropriately each point fits its assigned cluster compared to other clusters, helping assess the consistency and reliability of the clustering result.

•	Comparing clustering configurations: It can compare different clustering results or parameter settings within the same algorithm to select the best model.

•	Interpreting cluster assignments visually: Silhouette plots show the silhouette value for each point, helping identify poorly clustered points or clusters that may need refinement.

# 7.	What are the limitations of Hierarchical Clustering

key limitations of hierarchical clustering are:

•	Computationally expensive: Not suitable for large datasets due to high time and memory requirements.

•	Sensitive to noise and outliers: Can distort cluster formation.

•	Irreversible merges: Early wrong merges cannot be undone.

•	Choosing parameters is tricky: Distance metrics and linkage methods affect results.

•	Poor scalability: Struggles with high-dimensional or complex data.

•	Cluster shape assumptions: May not handle clusters of varying sizes or densities well.


# 8.	Why is feature scaling important in clustering algorithms like K—Means

Feature scaling is important in K-Means because it ensures all features contribute equally to distance calculations. Without scaling, features with larger ranges dominate, leading to biased and inaccurate clusters. Scaling improves clustering accuracy and helps the algorithm converge faster.

# 9.	How does DBSCAN identify noise points
-> DBSCAN identifies noise points as those that do not have enough neighbors within a specified radius (ε) to meet the minimum points (minPts) threshold for forming a dense region. Specifically, if a point’s ε-neighborhood contains fewer than minPts points, it is initially labeled as noise. However, this point can later be reclassified as a border point if it falls within the ε-neighborhood of a core point. Points that remain outside any cluster after this process are considered noise or outliers.

# 10.	Define inertia in the context of K-Means.
-> Inertia in K-Means clustering measures how well the data is clustered by calculating the sum of squared distances between each data point and its cluster centroid. Lower inertia indicates tighter, more compact clusters. It is used to evaluate clustering quality and to help choose the optimal number of clusters, often via the elbow method where inertia reduction slows as cluster count increases.

# 11. What is the elbow method in K-Means clustering?

-> The elbow method is a graphical technique used in K-Means clustering to find the optimal number of clusters (K). It involves plotting the within-cluster sum of squares (WCSS)-the sum of squared distances between data points and their cluster centroids-against different values of K. The plot typically shows a sharp decrease in WCSS as K increases, but after a certain point, the rate of decrease slows down, forming an "elbow" shape. The K at this elbow point is chosen as the optimal number of clusters because adding more clusters beyond it yields diminishing returns in reducing WCSS.


# 12.	Describe the concept of "density" in DBSCAN.

-> In DBSCAN, density refers to how closely packed data points are within a region. A region is considered dense if it contains at least a minimum number of points (minPts) within a specified radius (ε). DBSCAN forms clusters by connecting these dense regions, grouping points that have many neighbors nearby, while points in low-density areas are treated as noise or outliers. Essentially, density measures how many points lie close enough to each other to form a cluster.

#13.	Can hierarchical clustering be used on categorical data?

-> Yes, hierarchical clustering can be used on categorical data, but it requires defining an appropriate distance or dissimilarity measure suited for categorical variables since standard metrics like Euclidean distance don’t apply. By calculating a dissimilarity matrix using methods such as matching dissimilarity or Gower distance, hierarchical clustering (especially agglomerative) can be performed on purely categorical or mixed data types. Specialized functions and libraries (e.g., k-prototypes in Python) facilitate this process by computing distances and enabling linkage for categorical data clustering.

#14.	What does a negative Silhouette Score indicate?
-> A negative Silhouette Score indicates that a data point is likely assigned to the wrong cluster because it is closer, on average, to points in a different cluster than to points in its own cluster. This suggests poor clustering quality or overlapping clusters, and that the clustering configuration may be inappropriate for the data. In other words, negative values mean misclassification or that the data point fits better in another cluster.

#15.	Explain the term "linkage criteria" in hierarchical clustering.

-> In hierarchical clustering, linkage criteria define how the distance between two clusters is measured to decide which clusters to merge at each step. They specify how to calculate the dissimilarity between sets of points based on pairwise distances, shaping the cluster formation and final structure.

Common linkage methods include:

•	Single linkage: Distance between clusters is the shortest distance between any two points (one from each cluster). It tends to produce elongated clusters but is sensitive to noise.

•	Complete linkage: Distance is the largest distance between any two points in the clusters, leading to compact, well-separated clusters.

•	Average linkage: Distance is the average of all pairwise distances between points in the two clusters, balancing between single and complete linkage.

•	Ward’s method: Merges clusters to minimize the increase in total within-cluster variance, producing compact and similarly sized clusters.

#16.	Why might K-Means clustering perform poorly on data with varying cluster sizes or densities

-> K-Means clustering performs poorly on data with varying cluster sizes or densities because it assumes clusters are roughly spherical and of similar size. The algorithm assigns points based on distance to centroids, so clusters with larger sizes or lower densities can be split or merged incorrectly, leading to inaccurate clusters. Additionally, K-Means tends to favor clusters of similar density and size, struggling to capture clusters that differ significantly in these aspects. This limitation arises because K-Means minimizes variance within clusters but does not account for density variations or cluster shape irregularities.

# 17.	What are the core parameters in DBSCAN, and how do they influence clustering.

The core parameters in DBSCAN are:

•	ε (epsilon): Defines the maximum distance between two points for them to be considered neighbors. Smaller ε values create tighter, smaller clusters and may label more points as noise, while larger ε values produce larger clusters but risk merging distinct groups.

•	minPts (minimum points): The minimum number of points required within the ε-radius neighborhood to form a dense region (core point). Higher minPts values make the clustering stricter, requiring denser clusters, while lower values allow sparser clusters but may increase noise sensitivity.

# 18.	How does K—Means+ + improve upon standard K— Means initialization?

-> K-Means++ improves upon standard K-Means initialization by selecting initial centroids more strategically to spread them out across the data. Instead of choosing all centroids randomly, K-Means++

1.	Picks the first centroid randomly from the data points.

2.	For each subsequent centroid, selects a new point with probability proportional to the squared distance from the nearest existing centroid, ensuring centroids are far apart.

This approach leads to better initial cluster centers, which helps K-Means converge faster and often results in more accurate and stable clustering compared to random initialization. It reduces the chance of poor clustering due to unlucky initial centroid placement and improves both speed and quality of the final clusters

# 19.	What is agglomerative clustering?

-> Agglomerative clustering is a hierarchical, bottom-up clustering method that starts with each data point as its own cluster. It then repeatedly merges the two closest clusters based on a similarity or distance measure until all points are grouped into a single cluster or a stopping criterion is met. This process creates a tree-like structure called a dendrogram, showing the nested cluster relationships at different levels. Agglomerative clustering is widely used for its simplicity and ability to identify small clusters without needing to specify the number of clusters upfront.

#20.	What makes Silhouette Score a better metric than just inertia for model evaluation?

-> The Silhouette Score is considered better than inertia for clustering evaluation because it measures both cohesion (how close points are within the same cluster) and separation (how distinct clusters are from each other), providing a more balanced and nuanced assessment of cluster quality. In contrast, inertia only measures cluster compactness by summing squared distances within clusters and tends to decrease as the number of clusters increases, which can be misleading. The Silhouette Score ranges from -1 to 1, where higher values indicate well-separated, well-formed clusters, making it especially useful when the number of clusters is unknown or when you want to validate cluster consistency.








```python
# 21. Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize using a scatter plot.

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=50)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', s=200, marker='X')
plt.title("K-Means Clustering with 4 Centers")
plt.show()
```

```python
# 23. Generate synthetic data using make_moons and apply DBSCAN. Highlight outliers in the plot.

from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import numpy as np

X_moons, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels_dbscan = dbscan.fit_predict(X_moons)

plt.scatter(X_moons[:, 0], X_moons[:, 1], c=labels_dbscan, cmap='viridis', s=50)
outliers = labels_dbscan == -1
plt.scatter(X_moons[outliers, 0], X_moons[outliers, 1], c='red', s=100, label='Outliers', edgecolors='k')
plt.title("DBSCAN on make_moons with Outliers Highlighted")
plt.legend()
plt.show()
```

```python
# 24. Load the Wine dataset and apply K-Means clustering after standardizing the features. Print the size of each cluster.

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from collections import Counter

wine = load_wine()
X_wine = wine.data
scaler = StandardScaler()
X_wine_scaled = scaler.fit_transform(X_wine)

kmeans_wine = KMeans(n_clusters=3, random_state=42)
labels_wine = kmeans_wine.fit_predict(X_wine_scaled)

sizes = Counter(labels_wine)
print("Cluster sizes in Wine dataset:")
for cluster_id, size in sizes.items():
    print(f"Cluster {cluster_id}: {size} samples")
```

```python
# 29. Create synthetic data using make_blobs and evaluate silhouette scores for k=2 to 5. Display as a bar chart.

from sklearn.metrics import silhouette_score
import numpy as np

X_blob, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)
scores = []
ks = range(2, 6)

for k in ks:
    km = KMeans(n_clusters=k, random_state=42)
    labels = km.fit_predict(X_blob)
    score = silhouette_score(X_blob, labels)
    scores.append(score)

plt.bar(ks, scores, color='skyblue')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores for k=2 to 5')
plt.show()
```

```python
# 30. Load the Iris dataset and use hierarchical clustering to group data. Plot a dendrogram with average linkage.

from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import load_iris

iris = load_iris()
X_iris = iris.data

linked = linkage(X_iris, method='average')

plt.figure(figsize=(10, 7))
dendrogram(linked, labels=iris.target, leaf_rotation=90)
plt.title("Dendrogram (Average Linkage) - Iris Dataset")
plt.xlabel("Sample index")
plt.ylabel("Distance")
plt.show()
```

```python
# 31. Generate synthetic data with overlapping clusters using make_blobs, apply K-Means and visualize with decision boundaries.

from matplotlib.colors import ListedColormap
import numpy as np

X_overlap, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.5, random_state=42)
kmeans_overlap = KMeans(n_clusters=3, random_state=42)
labels_overlap = kmeans_overlap.fit_predict(X_overlap)

# Plot decision boundaries
h = 0.02
x_min, x_max = X_overlap[:, 0].min() - 1, X_overlap[:, 0].max() + 1
y_min, y_max = X_overlap[:, 1].min() - 1, X_overlap[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
Z = kmeans_overlap.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
plt.contourf(xx, yy, Z, cmap=cmap_light, alpha=0.5)
plt.scatter(X_overlap[:, 0], X_overlap[:, 1], c=labels_overlap, cmap='viridis', edgecolor='k', s=50)
plt.scatter(kmeans_overlap.cluster_centers_[:, 0], kmeans_overlap.cluster_centers_[:, 1], c='red', s=200, marker='X')
plt.title("K-Means with Decision Boundaries on Overlapping Clusters")
plt.show()
```
